**Nota Importante:** Este notebook contiene scripts de prueba diseñados exclusivamente para verificar que la instalación de las librerías y la configuración del hardware (CPU/GPU) se han realizado correctamente.

### Verificación de TensorFlow (funciona para los 3 sistemas operativos)

```python
import tensorflow as tf

# Listar dispositivos físicos (CPU)
cpus = tf.config.list_physical_devices('CPU')
print(f"CPUs disponibles: {len(cpus)}")

# Listar dispositivos físicos (GPU)
gpus = tf.config.list_physical_devices('GPU')

if gpus:
    print("\nTensorFlow está usando la GPU.")
    for gpu in gpus:
        print("Detalles del dispositivo:", gpu)
else:
    print("\nSolo se está utilizando la CPU. Algo salió mal")
```

### Verificación de PyTorch con CUDA (Windows/Linux con NVIDIA)

```python
import torch

if torch.cuda.is_available():
    print("PyTorch está usando la GPU (CUDA).")
    print("Nombre del dispositivo:", torch.cuda.get_device_name(0))
    
    # Asignación de un tensor a la memoria de la GPU
    device = torch.device('cuda')
    x = torch.randn(3, 3, device=device)
    print("\nTensor de prueba en CUDA:")
    print(x)
else:
    print("CUDA no está disponible en este equipo. PyTorch usará la CPU.")
```

### Verificación de PyTorch con MPS (Mac con Apple Silicon)

```python
import torch

if torch.backends.mps.is_available() and torch.backends.mps.is_built():
    print("PyTorch está usando la GPU de Apple (MPS).")
    
    # Asignación de un tensor a la memoria unificada de la Mac
    device = torch.device('mps')
    x = torch.randn(3, 3, device=device)
    print("\nTensor de prueba en MPS:")
    print(x)
else:
    print("MPS no está disponible en este equipo. PyTorch usará la CPU.")
```
